```python

import numpy as np
import matplotlib.pyplot as plt

# Vector operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vector Addition:", a + b)
print("Vector Multiplication:", a * b)

# Matrix operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix Addition:\n", A + B)
print("Matrix Multiplication:\n", np.dot(A, B))

# Plot vector
plt.plot(a, marker='o')
plt.title("Vector Visualization")
plt.xlabel("Index")
plt.ylabel("Value")
plt.show()
```
